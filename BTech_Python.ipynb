# Sets

A set is an unordered collection of items. Every set element is unique (no duplicates) and must be immutable (cannot be changed).

However, a set itself is mutable. We can add or remove items from it.

Sets can also be used to perform mathematical set operations like union, intersection, symmetric difference, etc.

Characterstics:
- Unordered
- Mutable
- No Duplicates
- Can't contain mutable data types

### `*args and **kwargs`
`*args and **kwargs` are special Python keywords that are used to pass the variable length of arguments to a function

````python
# *args
# allows us to pass a variable number of non-keyword arguments to a function.

def multiply(*salmaan):
  print(type(salmaan))
  mult=1
  for i in salmaan:
    mult=mult*i
  return mult

multiply(1,2,3,4,5)
````

````python
# **kwargs
# **kwargs allows us to pass any number of keyword arguments.
# Keyword arguments mean that they contain a key-value pair, like a Python dictionary.

def display(**args):
  print(type(args))
  for i,j in args.items():
    print(i,"--->",j)
  for i in args:
    print(i,"--->",args[i])




````

````python
class atm:
  def __init__(self):
    self.pin=1234
    self.balance=1000
    self.L=[]
    self.menu()

  def showmini(self):
    history=self.L[::-1]
    for i in history:
      print(i)


  def showbalance(self):

    pinn=int(input("ENter the pin"))
    if pinn==self.pin:
      print("Balance--->", self.balance)


  def deposit(self):
    pinn=int(input("ENter the pin"))
    if pinn==self.pin:

      amount=int(input("enter the amount"))
      if amount>=0:
        self.balance+=amount
        self.L.append(['+', amount])
        print("Updated balance-->", self.balance)
      else:
        print("Jyada hoshiyaar hai kya!!!")
      self.menu()

  def withdrawl(self):
    pinn=int(input("ENter the pin"))
    if pinn==self.pin:

      amount=int(input("enter the amount"))
      if amount>=0 and amount<=self.balance:
        self.balance-=amount
        self.L.append(['-', amount])
        print("Updated balance-->", self.balance)
      else:
        print("Bhai tu garib hai!!!")

      self.menu()

  def change_pin(self):
    pinn=int(input("Enter the default pin"))
    if pinn==self.pin:
      new_pin=int(input("ENter new pin"))
      self.pin=new_pin
      print("Pin changed successfully")
    else:
      print(" Default pin galat hai Bro!!!")

    self.menu()

  def menu(self):
    print("""
    *************************
    Press 1 for pin change
    Press 2 for deposit
    Presee 3 for Withdrawl
    Press 4 for Show Balance
    Press 5 for Mini Statement
    Press 6 for Exit
    *************************
    """)
    user_input=int(input("enter the choice"))
    if user_input==1:
      self.change_pin()

    if user_input==2:
      self.deposit()

    if user_input==3:
      self.withdrawl()

    if user_input==4:
      self.showbalance()

    if user_input==5:
      self.showmini()

    if user_input==6:
      exit(0)


a=atm()
````

### Static methods

##### Points to remember about static

- Static attributes are created at class level.
- Static attributes are accessed using ClassName.
- Static attributes are object independent. We can access them without creating instance (object) of the class in which they are defined.
- The value stored in static attribute is shared between all instances(objects) of the class in which the static attribute is defined.

````python
class Lion:
  __water_source="well in the circus"

  def __init__(self,name, gender):
      self.__name=name
      self.__gender=gender

  def drinks_water(self):
      print(self.__name,
      "drinks water from the",Lion.__water_source)

  @staticmethod
  def get_water_source():
      return Lion.__water_source

simba=Lion("Simba","Male")
simba.drinks_water()
print( "Water source of lions:",Lion.get_water_source())
````
